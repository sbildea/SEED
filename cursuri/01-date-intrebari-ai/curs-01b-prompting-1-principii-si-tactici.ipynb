## Principii de prompting
- **Principiul 1: Scrie instrucțiuni clare și specifice**
- **Principiul 2: Lasă modelul să “gândească”**

### Tactici

#### Tactica 1: Folosește delimitatoare pentru a indica clar părțile distincte ale intrării
- Delimitatoarele pot fi orice, de ex.: ```, """, < >, `<tag> </tag>`, `:`

````python
text = f"""
Ar trebui să exprimi ce vrei să facă un model oferind instrucțiuni cât se poate de clare și specifice. 
Acest lucru va ghida modelul către rezultatul dorit, 
și va reduce șansele de a primi răspunsuri irelevante 
sau incorecte. Nu confunda scrierea unui 
prompt clar cu scrierea unui prompt scurt. 
În multe cazuri, prompturile mai lungi oferă mai multă claritate 
și context pentru model, ceea ce poate duce la 
rezultate mai detaliate și relevante.
"""
prompt = f"""
Rezumă textul delimitat de triple backticks 
într-o singură propoziție cat mai scurtă.
```{text}```
"""

print(prompt)
response = get_completion(prompt = prompt)
print('Răspuns:\n', response)
````

### Principiul 2: Lasă modelul să “gândească”

````python
text = f"""
Într-un sat fermecător, frații Jack și Jill au plecat într-o 
aventură să aducă apă de la un izvor de pe deal. Pe măsură ce 
urcau, cântând vesel, ghinionul a lovit — Jack s-a împiedicat de 
o piatră și a căzut pe deal, iar Jill l-a urmat. 
Deși ușor răniți, cei doi s-au întors acasă la îmbrățișări 
mângâietoare. În ciuda neplăcerii, spiritul lor aventurier a rămas 
neclintit și au continuat să exploreze cu bucurie.
"""
# exemplul 1
prompt_1 = f"""
Efectuează următoarele acțiuni: 
1 - Rezumă următorul text delimitat de triple backticks într-o propoziție.
2 - Tradu rezumatul în limba engleza.
3 - Listează fiecare nume din rezumatul în limba engleza.
4 - Oferă un obiect JSON care conține următoarele chei: english_summary, num_names.

Separă răspunsurile cu linii noi.

Text:
```{text}```
"""
print(prompt_1)
````

#### Tactica 2: Instruiește modelul să-și găsească propria soluție înainte de a trage o concluzie

````python
prompt = f"""
Determină dacă soluția studentului este corectă sau nu.

Întrebare:
Construiesc o instalație solară și am nevoie de ajutor 
pentru a calcula finanțele. 
- Costul terenului este de 100 $ / metru pătrat
- Pot cumpăra panouri solare cu 250 $ / metru pătrat
- Am negociat un contract pentru întreținere care va costa 
  o sumă fixă de 100k $ pe an și încă 10 $ / metru pătrat
Care este costul total pentru primul an de funcționare 
în funcție de numărul de metri pătrati.

Soluția studentului:
Fie x dimensiunea instalației în metri pătrati.
Costuri:
1. Costul terenului: 100x
2. Costul panourilor solare: 250x
3.Costul întreținerii: 100.000 + 100x 
Cost total: 100x + 250x + 100.000 + 100x = 450x + 100.000
"""

print(prompt)
````

````python

response = get_completion(prompt=prompt)
print(response)
````

Soluția studentului este **incorectă**.

**Motivare:**
Eroarea este în calculul costului de întreținere. Problema spune că costul de întreținere este "$100k pe an, plus încă **$10** / metru pătrat."

Studentul a folosit greșit **$100 / metru pătrat** pentru partea variabilă a costului de întreținere în loc de **$10 / metru pătrat** specificat în prompt.

**Calcul corect:**
1. Costul terenului: $100x$
2. Costul panourilor solare: $250x$
3. Costul întreținerii: $100,000 + 10x$

**Funcția cost total:**
$100x + 250x + 10x + 100,000 = \mathbf{360x + 100,000}$

- În curs, folosim un backslash `\` pentru a face textul să se potrivească pe ecran fără a introduce caractere de newline `\n`.

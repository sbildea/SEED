````python
prompt = """
Sarcina ta este să generezi un rezumat scurt al unei recenzii
de produs de pe un site de comerț electronic.

Rezumă recenzia de mai jos, delimitată de trei backticks,
în cel mult 20 de cuvinte. Adauga un json cu urmatoarele chei: "pret", "valoare_perceputa", "comentariu".
"""
for i in range(len(reviews)):
    
    prompt += f""" Recenzie{i}: ```{reviews[i]}```
    """

print(prompt)
````

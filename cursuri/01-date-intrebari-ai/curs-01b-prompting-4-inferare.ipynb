## Creează o alertă de știri pentru anumite subiecte

```python
prompt = f"""
Determină dacă fiecare element din următoarea listă de 
subiecte este un subiect în textul de mai jos, care este 
delimitat de triple backticks.

Listă de subiecte: {", ".join(topic_list)}
Raspunsul sa fie un obiect JSON cu cheile fiind subiectele și valorile fiind "yes" sau "no".
Text: '''{story}'''
"""

print(prompt)
```

```python

response = get_completion(prompt=prompt)
print(response)
```

Pe baza textului furnizat, iată determinarea pentru fiecare subiect:

*   **nasa**: Da (Textul discută în mod explicit despre NASA, evaluarea sa ridicată a satisfacției și managementul său).
*   **guvern local**: Nu (Textul se referă la "guvern" și departamente federale precum Social Security Administration, dar nu menționează guvernul local).
*   **inginerie**: Nu (Deși textul menționează munca "inovatoare", nu folosește explicit cuvântul "inginerie" și nu descrie taskuri de inginerie).
*   **satisfacția angajaților**: Da (Textul se concentrează pe un sondaj privind nivelul de satisfacție al angajaților din sectorul public).
*   **guvern federal**: Da (Textul menționează sondaje realizate de guvern și departamente federale precum NASA și Social Security Administration).

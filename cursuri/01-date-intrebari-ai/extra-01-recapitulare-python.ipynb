# Excepţii
Când ceva nu merge bine, Python ridică o excepție. Dacă acestea nu vor fi manipulate
programul tău poate să clacheze. Le poți gestiona folosind try și except:

```python
try:
    print (0 / 0)
except ZeroDivisionError:
    print ("nu se poate împărți la zero")
```


Deși în multe limbi, excepțiile sunt considerate rele, în Python nu există nicio rușine
în utilizarea acestora pentru a vă face mai curat codul și vom face acest lucru ocazional.

```python
try:
    print (0 / 0)
except ZeroDivisionError as exception:
    print ("nu se poate împărți la zero:", exception)
```

Puteți obține sau seta al n -lea element al unei liste cu paranteze pătrate:

## defaultdict

Imaginează-ți că încerci să numeri cuvintele și aparițiile lor dintr-un document. O abordare evidentă este
să creezi un dicționar în care cheile sunt cuvinte și valorile sunt numărul de apariții. Pe măsură ce verificați
fiecare cuvânt, puteți crește numărul său dacă este deja în dicționar și îl puteți adăuga la
dicționar dacă nu este:


```python
document = 'Imaginează-ți că încerci să numeri cuvintele și aparițiile lor dintr-un document. \
O abordare evidentă este să creezi un dicționar în care cheile sunt cuvinte și valorile sunt numărul de apariții. \
Pe măsură ce verificați fiecare cuvânt, puteți crește numărul său dacă este deja în dicționar și îl puteți adăuga \
la dicționar dacă nu este:'.split(' ')
```

```python
word_counts = {}
for word in document:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1
print('cuvinte care apar cel puțin de 2 ori:')
[(w,word_counts[w]) for w in word_counts if word_counts[w] >1]
```

Puteți de asemenea să vă ocupați doar de excepția de la încercarea de a căuta o cheie care lipsește:

```python
word_counts = {}
for word in document:
    try:
        word_counts[word] += 1
    except KeyError:
        word_counts[word] = 1
```


O a treia abordare este de a utiliza get, care se comportă cu grație pentru cheile lipsă:

```python
word_counts = {}
for word in document:
    previous_count = word_counts.get(word, 0) # returnează direct 0 pentru chei negăsite
    word_counts[word] = previous_count + 1
```

Un dicționar predefinit (defaultdict) este ca un dicționar obișnuit, cu excepția cazului în care înceri să cauți o cheie pe care nu o conține: mai întâi adaugă o valoare folosind o funcție de argument zero pe care ai furnizat-o
când l-ai creat. Pentru a utiliza dicționare predefinite, trebuie să le importăm din modulul de
colecții:

```python
from collections import defaultdict
word_counts = defaultdict(int) # int() produce 0 impicit
for word in document:
    word_counts[word] += 1
    
word_counts
```


De asemenea, dicționarele predefinite pot fi utile în combinație cu listă sau dict sau chiar cu propriile funcții:

O problemă a listelor este că acestea pot crește ușor foarte mari. Intervalul (1000000) creează un

Un generator este ceva cu care puteți itera (pentru noi, de obicei folosind <b>for</b>), dar al cărui
valori sunt produse numai după cum este necesar (lazily /în mod leneș).
O modalitate de a crea generatoare este cu funcții și operatorul yield:

```python
def lazy_range(n):
    """o versiune lazy pentru range"""
    i = 0
    while i < n:
        yield i
        i += 1
```

# Programare orientată pe obiecte

Ca în multe alte limbaje de programare, Python vă permite să definiți clase care încapsulează date și
funcții care operează asupra lor. Le vom folosi uneori pentru a face codul nostru mai curat și
mai simplu. Este probabil cel mai simplu să le explici construind un material puternic adnotat
exemplu.
Imaginați-vă că nu am avut setul integrat Python. Atunci s-ar putea să vrem să ne creăm propria noastră
Set de clase.
Ce comportament ar trebui să aibă clasa noastră? Având în vedere o instanță a Setului, va trebui să fim capabili
adăugați elemente la el, eliminați elemente din acesta și verificați dacă conține o anumită valoare. Bine
creează toate acestea ca funcții de membru, ceea ce înseamnă că le vom accesa cu un punct după
Set obiect:

```python
# by convention, we give classes PascalCase names
class Set:

    # these are the member functions
    # every one takes a first parameter "self" (another convention)
    # that refers to the particular Set object being used
    
    def __init__(self, values=None):
        """This is the constructor.
        It gets called when you create a new Set.
        You would use it like
        s1 = Set() # empty set
        s2 = Set([1,2,2,3]) # initialize with values"""
        
        self.dict = {} # each instance of Set has its own dict property
        
        # which is what we'll use to track memberships
        
        if values is not None:
            for value in values:
                self.add(value)
    
    
    def __repr__(self):
        """this is the string representation of a Set object
        if you type it at the Python prompt or pass it to str()"""

        return "Set: " + str(self.dict.keys())
    
    
    # we'll represent membership by being a key in self.dict with value True
    def add(self, value):
    
        self.dict[value] = True
    
    
    # value is in the Set if it's a key in the dictionary
    def contains(self, value):
    
        return value in self.dict
    
    
    def remove(self, value):
        
        del self.dict[value]
```


Pe care am putea apoi să le folosim astfel:

```python

s = Set([1,2,3])
s.add(4)
print (s.contains(4)) # True
s.remove(3)
print (s.contains(3)) # False
```

# args și kwargs
Să zicem că vrem să creăm o funcție de ordin superior care să ia ca intrare unele funcții f și
returnează o nouă funcție care pentru orice intrare returnează de două ori valoarea f:

```python
def doubler(f):
    
    def g(x):
    
        return 2 * f(x)
    
    return g
```

```python

#Acest lucru funcționează în unele cazuri:
```

```python
def f1(x):
    return x + 1
g = doubler(f1)
print (g(3)) # 8 (== ( 3 + 1) * 2)
print (g(-1)) # 0 (== (-1 + 1) * 2)
```


Cu toate acestea, nu mai merge cu funcțiile care au mai mult de un singur argument:

```python

def f2(x, y):
    return x + y

g = doubler(f2)
print (g(1, 2)) # TypeError: g() takes exactly 1 argument (2 given)
```


Ceea ce avem nevoie este un mod de a specifica o funcție care să ia argumente arbitrare. Noi putem sa facem asta
cu dezlegarea argumentelor și un pic de magie:

```python
def magic(*args, **kwargs):
    print ("unnamed args:", args)
    print ("keyword args:", kwargs)
magic(1, 2, key="word", key2="word2")
# prints
# unnamed args: (1, 2)
# keyword args: {'key2': 'word2', 'key': 'word'}
```

Adică, atunci când definim o funcție ca aceasta, args este un tuplu al argumentelor sale nenumite și
kwargs este un dict al argumentelor sale numite. Funcționează și în alt mod, dacă doriți să utilizați 
lista (sau tuple) și dict pentru a furniza argumente unei funcții:

```python
def other_way_magic(x, y, z):
    return x + y + z
x_y_list = [1, 2]
z_dict = { "z" : 3 }
print (other_way_magic(*x_y_list, **z_dict)) # 6
```


Ați putea face tot felul de trucuri ciudate cu asta; îl vom folosi doar pentru a produce o comandă superioară
funcții ale căror intrări pot accepta argumente arbitrare:

```python
def doubler_correct(f):
    
    """works no matter what kind of inputs f expects"""
    def g(*args, **kwargs):
        """whatever arguments g is supplied, pass them through to f"""
        return 2 * f(*args, **kwargs)
    
    return g

g = doubler_correct(f2)
print (g(1, 2)) # 6
```

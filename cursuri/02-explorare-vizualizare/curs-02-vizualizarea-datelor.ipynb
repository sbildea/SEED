#### 2. Demonstrarea valorii adăugate printr-un studiu folosind deviația standard

În statistica descriptivă întâlnim doi indicatori ai împrăștierii - dispersia și deviația standard: 
- utile pentru repartițiile relativ simetrice față de medie 
- măsoară deviația față de medie

În cazul unei variabile aleatoare discrete cu $N$ valori posibile:

$\sigma = \sqrt{\frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^2},$ unde $\mu = \frac{1}{N}\sum_{i=1}^{N}x_i.$

În cazul unui eșantion de mărime $N$ cu media $\bar{x}$, formula este
$s = \sqrt{\frac{1}{N-1}\sum_{i=1}^{N}(x_i-\bar{x})^2}$.

`Atenție` Dacă se cunoaște media $\mu$ a populației din care provine eșantionul cu media $\bar{x}$ nu e nevoie de corecția $N-1$ în loc de $N$ și se folosește formula cu media populației pe datele eșantionului.

##### 3.5 Distribuții și boxplot în Python

Până aici am văzut ce vizualizări folosesc firmele. Cum le facem noi? Trei biblioteci: `matplotlib` (baza), `seaborn` (grafice statistice direct din DataFrame) și `plotly` (interactiv). Regula de lucru: **întâi întrebarea, apoi graficul**.

Folosim prețurile caselor din Ames, Iowa (`lab-set6`, 1 460 de case). Întrebarea: *cât costă o casă „obișnuită” și cât de mult se abat prețurile de la ea?*

```python
import seaborn as sns
import matplotlib.pyplot as plt
from seed.date import incarca

case = incarca("lab-set6")[1]          # tabelul cu prețurile caselor
pret = case["SalePrice"]

# histogramă: împarte valorile în intervale și numără câte case cad în fiecare; kde = curba de densitate
sns.histplot(pret, kde=True)
plt.title("Prețul caselor: majoritatea sub 250 000, cu o coadă spre scump")
plt.show()
print("medie:", round(pret.mean()), " mediană:", pret.median())
```

**Cum citim histograma.** Înălțimea barei = câte case au prețul în acel interval; curba (*densitatea*) este versiunea netezită. Forma ne spune mai mult decât media:

- distribuția are **coadă spre dreapta** (asimetrie pozitivă): câteva case foarte scumpe trag media (≈ 180 900) peste mediană (163 000);
- în astfel de cazuri **mediana** descrie mai bine „casa obișnuită”.

`Q&A` Dacă ați raporta managerului doar media, ce impresie falsă i-ați crea?

```python
# boxplot: aceeași variabilă, rezumată prin cuartile
q1, mediana, q3 = pret.quantile([0.25, 0.5, 0.75])
iqr = q3 - q1                       # IQR = Q3 - Q1: lățimea cutiei, adică intervalul în care se află jumătatea din mijloc
limita_sus = q3 + 1.5 * iqr         # peste această valoare punctele sunt marcate ca „aberante”
print(q1, mediana, q3, iqr, limita_sus, (pret > limita_sus).sum())

sns.boxplot(x=pret)
plt.title("Boxplot: cutia = Q1-Q3, linia = mediana, punctele = peste Q3 + 1,5·IQR")
plt.show()
```

**Cum citim boxplot-ul.** Cutia cuprinde 50% dintre case (de la Q1 la Q3), linia din cutie este mediana, iar mustățile se opresc la ultima valoare aflată la cel mult 1,5 × IQR de cutie. Punctele de dincolo sunt **posibile** valori aberante: aici 61 de case scumpe. Nu sunt neapărat erori: decizia de a le păstra sau elimina depinde de întrebare.

Puterea boxplot-ului apare la **comparație**: mai multe distribuții alăturate, pe aceeași scală.

```python
# aceeași variabilă, comparată pe categorii: case cu și fără aer condiționat central
sns.boxplot(x="CentralAir", y="SalePrice", data=case)
plt.title("Casele cu aer condiționat central costă mai mult")
plt.show()
print(case.groupby("CentralAir")["SalePrice"].median())
```
